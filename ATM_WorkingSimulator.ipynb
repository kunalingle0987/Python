```python
def showbalance(balance):
    print("Your current balance is", balance)

def deposite(balance):
  amount=int(input("Enter the amount you want to deposite"))
  if amount>0:
    balance=balance+amount
    print("successfully Deposit Amount!")
    print("Your current balance is", balance)
  else:
    print("Invalid amount")
  return balance

def withdraw(balance):
  try:
    amount=int(input("Enter the amount you want to withdraw"))
    if amount>balance:
      print("insufficient Balance")
    elif amount<=0:
      print("invalid amount")
    else:
      balance=balance-amount
      print("sucessfully withdraw amount")
      print("Your current balance is: ",balance)
  except ValueError:
    print("Invalid num ! Enter proper value")

  return balance


def main():
  balance=10000
  is_running=True


  print("||WELCOME TO ATM ||")

  while is_running:
    print("\n Please choose option:")
    print("1. Check Balance")
    print("2. Deposite Money")
    print("3. Withdraw money")
    print("4. Exit")

    choice=input("Enter your choice: ")
    if choice=="1":
      showbalance(balance)
    elif choice=="2":
     balance=deposite(balance)
    elif choice=="3":
      balance=withdraw(balance)
    elif choice=="4":
      is_running=False
      print("Thank You Visit Again!")
    else:
      print("Invalid choice choose a valid num...")

if __name__=="__main__":
  main()
```
